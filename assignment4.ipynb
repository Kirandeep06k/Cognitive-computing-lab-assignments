```python
#Q1
import pandas as pd
rno = 1024170280
d1 = ( rno % 100 ) // 10
d2 = rno % 10

categories = ["billing", "account", "general"]

category1 = categories[d1 % 3]
category2 = categories[d2 % 3]


fixed_entries = [ {"question": "what is the annual fee", 
                   "answer": "The annual fee is Rs 500.", 
                   "keywords": "fee cost price charge", 
                   "category": "billing"}, 
                 
                 {"question": "how to reset password",
                  "answer": "Go to Settings > Reset Password.", 
                  "keywords": "password reset login", 
                  "category": "account"}, 
                 
                 {"question": "what are your working hours", 
                  "answer": "We are open 9 AM to 5 PM.", 
                  "keywords": "hours timing open time", 
                  "category": "general"}, 
                 
                 {"question": "how can i pay the fee",
                  "answer": "You can pay via UPI, card, or net banking.", 
                  "keywords": "pay payment upi fee",
                  "category": "billing"}, ] 

e5 = {
    "question": "how do i file a complaint",
    "answer": "You can check the official website",
    "keywords": "complaint issue problem file",
    "category": category1
}

e6 = {
    "question": "how can i check my payment status",
    "answer": "You can check your payment status in the Billing section.",
    "keywords": "payment status billing",
    "category": category2
}

final = fixed_entries + [e5, e6]
df = pd.DataFrame(final)
print(df)
```

```python
#Q2
def score_query(query):
    results = []
    
    for i in range(len(df)):
        score = 0

        keywords = df.loc[i, "keywords"].split()

        for word in keywords:
            if word.lower() in query.lower():
                score = score + 1

        if score > 0:
            results.append((df.loc[i], score))

    return results


query = input("Enter your query: ")

results = score_query(query)

for entry, score in results:
    print("\nQuestion:", entry["question"])
    print("Answer:", entry["answer"])
    print("Category:", entry["category"])
    print("Confidence:", score)
```

```python
#Q3
def same_category(category_name, df):
    results = []
    for i in range(len(df)):
        if df.loc[i, "category"] == category_name:
            results.append(df.loc[i, "question"])
    return results

# e5's category is category1 = "general"
matches = same_category(category1, df)
print(matches)
```
