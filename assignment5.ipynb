```python
import numpy as np
# Q1
a = np.array([0,1,2,3,4])
print("Original array ", a)
# a
print("Add 2 ", a + 2)
# b
print("Mul by 3 ", a * 3)
# c
print("Div by 2 ", a / 2)
```

```python
# Q2 
# a
arr = np.array([1, 2, 3, 6, 4, 5])
r = np.flip(arr)
print("Revered arr : ", r)
# b
x = np.array([1,2,3,4,5,1,2,1,1,1]) 
# i
mfreq =  np.bincount(x).argmax() 
ind = np.where(x == mfreq)[0]
print("Most freq value = ", mfreq, " and index = ", ind)

# ii 
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
mfreq2 =  np.bincount(y).argmax() 
ind2 = np.where(y == mfreq2)[0]
print("Most freq value = ", mfreq2, " and index = ", ind2)
```

```python
# Q4
Kiran = np.linspace(10, 100, 25)
print(Kiran.ndim)
print(Kiran.shape)
print(Kiran.size)
print(Kiran.dtype)
print(Kiran.nbytes)

t = Kiran.reshape(25, 1)
print(t)
print(Kiran.T)  # no change for 1-D array
```

```python
# Q5
ucs420_kiran = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(ucs420_kiran.mean())
print(np.median(ucs420_kiran))
print(ucs420_kiran.max())
print(ucs420_kiran.min())
print(np.unique(ucs420_kiran))

reshaped_ucs420_kiran = ucs420_kiran.reshape(4, 3)
print(reshaped_ucs420_kiran)

resized_ucs420_kiran = np.resize(ucs420_kiran, (2, 3))
print(resized_ucs420_kiran)
```
