```python
# Q3
import random

# Use your own roll number
roll_number = 1024170280

# i
random.seed(roll_number)
numbers = []
for i in range(100):
    numbers.append(random.randint(100, 900))
print("Random numbers:", numbers)
```

```python

# ii 
odd = []
for num in numbers:
    if num % 2 != 0:
        odd.append(num)

print("Odd numbers:", odd)
print("Number of odd numbers:", len(odd))
```

```python
# iii
even = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)

print("Even numbers:", even)
print("Number of even numbers:", len(even))
```

```python
# iv
prime = []

for num in numbers:
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

    if count == 2:
        prime.append(num)

print("Number of prime numbers:", len(prime))
# Prime numbers using list comprehension
prime_list = [num for num in numbers if num in prime]
print("Prime numbers:", prime_list)
```

```python
# v
most = numbers[0]
frequency = numbers.count(numbers[0])

for num in numbers:
    if numbers.count(num) > frequency:
        most = num
        frequency = numbers.count(num)

print("Most frequent number:", most)
print("Number of times it occurs:", frequency)
```

```python
# Q4

roll = "1024170280"

A = set()
B = set()

for digit in roll:
    digit = int(digit)

    A.add(digit * 7)
    B.add(digit * 9)

print("Set A:", A)
print("Set B:", B)

# vi. Union
union = A.union(B)
print("Union:", union)
```

```python

# vii
intersection = A.intersection(B)
print("Intersection:", intersection)
```

```python

# viii
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives values present in one set but not the other.
# symmetric_difference() gives values that are in either set but not in both.
```

```python

# ix
symmetric = A.symmetric_difference(B)

print("Symmetric difference:", symmetric)
```

```python
# x
print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))
```

```python
# xi
X = int(input("Enter a value: "))
A.discard(X)
print("Set A after discard:", A)
# discard() does not give an error if X is not present.
# remove() gives an error if X is not present.
```

```python
# Q5

my_dict = {
    "name": "Kirandeep Kaur",
    "roll_no": "1024170280",
    "branch": "CSE",
    "age": 20,
    "city": "Samrala"
}
print("Original dictionary:", my_dict)

# i
value = my_dict.pop("city")
my_dict["location"] = value

print("After renaming city:", my_dict)
```

```python
# ii
my_dict["cgpa"] = 8.51

print("After adding CGPA:", my_dict)
```

```python
# iii
my_dict["age"] = my_dict["age"] + 1

print("After increasing age:", my_dict)
```

```python

# iv
dict1 = my_dict.copy()

removed = dict1.pop("branch")

print("Using pop():", dict1)
print("Removed value:", removed)


dict2 = my_dict.copy()

del dict2["branch"]

print("Using del:", dict2)

# pop() removes the key and returns its value, while del only removes the key.
```

```python

# v
for key, value in my_dict.items():
    print(key, "→", value)
```

```python
# vi
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present.")
```

```python
# vii
friend_dict = {
    "name": "Himan",
    "roll_no": "123456",
    "branch": "Biotech",
    "age": 21,
    "city": "Ludhiana"
}
merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:", merged_dict)

# If both dictionaries have the same key, the value from the second dictionary wins.
```

```python
# viii
string_dict = {key: value for key, value in my_dict.items() if type(value) == str}

print("Dictionary with string values:", string_dict)
```
